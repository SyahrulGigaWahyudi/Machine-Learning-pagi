```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns

#membaca file csv menggunakan pandas
df = pd.read_csv("/content/drive/MyDrive/Pratikum_Ml/pratikum06/data/breast_cancer.csv")

#cetak headet data (baris data) dari file
df.head()

```

```python
df.info()
df.describe()
df['diagnosis'].value_counts()
```

```python
df = df.drop(columns=['id'])
le = LabelEncoder()
df['diagnosis_encoded'] = le.fit_transform(df['diagnosis'])
```

```python
feature_columns = ['radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean']
X = df[feature_columns]
y = df['diagnosis_encoded']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
svc_linear = SVC(kernel='linear', probability=True, random_state=42)
svc_rbf    = SVC(kernel='rbf', probability=True, random_state=42)

svc_linear.fit(X_train_scaled, y_train)
svc_rbf.fit(X_train_scaled, y_train)
```

```python
y_pred_linear = svc_linear.predict(X_test_scaled)
y_pred_rbf    = svc_rbf.predict(X_test_scaled)

print(accuracy_score(y_test, y_pred_rbf))
print(classification_report(y_test, y_pred_rbf))
```

```python
cm = confusion_matrix(y_test, y_pred_rbf)
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Benign','Malignant'],
            yticklabels=['Benign','Malignant'])
```

```python
# --- Visualisasi 3D Dataset Breast Cancer ---
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

# Ambil tiga fitur utama untuk divisualisasikan
X_3d = X_test[['radius_mean', 'texture_mean', 'perimeter_mean']]
y_3d = y_test.values

# Buat figure 3D
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Pisahkan data berdasarkan label
benign_idx = np.where(y_3d == 0)
malignant_idx = np.where(y_3d == 1)

# Plot masing-masing kelas dengan warna berbeda
ax.scatter(
    X_3d.iloc[benign_idx[0], 0],
    X_3d.iloc[benign_idx[0], 1],
    X_3d.iloc[benign_idx[0], 2],
    c='skyblue', label='Benign', s=40
)
ax.scatter(
    X_3d.iloc[malignant_idx[0], 0],
    X_3d.iloc[malignant_idx[0], 1],
    X_3d.iloc[malignant_idx[0], 2],
    c='salmon', label='Malignant', s=40
)

# Label sumbu
ax.set_xlabel('Radius Mean')
ax.set_ylabel('Texture Mean')
ax.set_zlabel('Perimeter Mean')

# Judul dan legenda
ax.set_title('Visualisasi 3D Data Breast Cancer (Test Set)')
ax.legend()
plt.tight_layout()
plt.show()
```
