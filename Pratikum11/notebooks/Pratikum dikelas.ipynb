```python
import matplotlib.pyplot as plt

# Membuat figure dan subplot 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Membuat scatter plot 3D
scatter = ax.scatter(
    df["SepalLengthCm"],
    df["PetalLengthCm"],
    df["PetalWidthCm"],
    c=df["Cluster"],
    s=60
)

# Mengatur judul dan label sumbu
ax.set_title("3D Plot K-Means Clustering pada Iris")
ax.set_xlabel("Sepal Length (cm)")
ax.set_ylabel("Petal Length (cm)")
ax.set_zlabel("Petal Width (cm)")

# Legend cluster
# Asumsi ada 3 cluster (0, 1, 2)
legend_labels = [f"Cluster {i+1}" for i in range(3)]
# Mengambil elemen legend pertama (handles) dan menggunakan labels yang dibuat
ax.legend(handles=scatter.legend_elements()[0], labels=legend_labels)

plt.show()
```
