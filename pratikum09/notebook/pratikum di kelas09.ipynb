```python
replacements = {
    'Sex':{'male': 0, 'female': 1},
    'Embarked': {'S': 0, 'C': 1, 'Q': 2}
}
titanic_data.replace(replacements, inplace=True)
```

```python
X = titanic_data.drop(columns=['Survived'])
Y = titanic_data['Survived']
```

```python
X.head()
```

```python
Y.head()
```

```python
# 1. Membagi data menjadi Training dan Testing set
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)
```

```python
# 2. Mengecek ukuran data (opsional, untuk memastikan pembagian benar)
print(X.shape, X_train.shape, X_test.shape)
```

```python
# 3. Melakukan Scaling (Standarisasi)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.naive_bayes import GaussianNB

# 1. Membuat model kosong menggunakan algoritma Naive Bayes
nb_model = GaussianNB()

# 2. Melatih model ("Fit") menggunakan data latihan yang sudah distandarisasi
# Model akan belajar pola hubungan antara fitur (X_train_scaled) dan target (Y_train)
nb_model.fit(X_train_scaled, Y_train)
```

```python
from sklearn.metrics import accuracy_score

# 1. Suruh model menebak jawaban (Predict)
# Menebak data latihan (Train)
train_pred_nb = nb_model.predict(X_train_scaled)
# Menebak data ujian (Test)
test_pred_nb = nb_model.predict(X_test_scaled)
```

```python
# 2. Cek Nilai Akurasi (Bandingkan tebakan model dengan kunci jawaban asli)
print("Training Accuracy (NB): ", accuracy_score(Y_train, train_pred_nb))
print("Testing Accuracy (NB): ", accuracy_score(Y_test, test_pred_nb))
```

```python
# Visualisasi Confusion Matrix (Naive Bayes)

# 1. Atur ukuran gambar (lebar 6, tinggi 4)
plt.figure(figsize=(6,4))

# 2. Hitung matriks kebingungan (Bandingkan Kunci Jawaban vs Tebakan Model)
cm_nb = confusion_matrix(Y_test, test_pred_nb)

# 3. Gambar Heatmap (Peta Panas)
# annot=True: Tampilkan angkanya
# fmt="d": Angka bilangan bulat (bukan desimal)
# cmap="Blues": Warna biru
sns.heatmap(cm_nb, annot=True, fmt="d", cmap="Blues",
            xticklabels=['Not Survived', 'Survived'],
            yticklabels=['Not Survived', 'Survived'])

# 4. Beri Judul dan Label
plt.title("Confusion Matrix - Naive Bayes")
plt.xlabel("Predicted") # Sumbu X = Tebakan Model
plt.ylabel("Actual")    # Sumbu Y = Kenyataan
plt.show()
```

```python
from sklearn.metrics import classification_report

print("\nClassification Report (NB):")
print(classification_report(Y_test, test_pred_nb))
```

```python
from sklearn.model_selection import cross_val_score

# 1. Lakukan Uji Silang (Cross Validation)
# nb_model : Model yang mau diuji
# cv=5     : Data diacak dan dibagi jadi 5 bagian (biar diuji 5 kali)
cv_nb = cross_val_score(nb_model, X, Y, cv=5, scoring='accuracy')

print("\nNaive Bayes Cross Validation Accuracy (5-Fold):")
print("Scores:", cv_nb)          # Nilai akurasi di tiap pengujian (ada 5 nilai)
print("Mean Accuracy:", cv_nb.mean()) # Rata-rata dari 5 nilai tersebut
print("Std Deviation:", cv_nb.std())  # Standar deviasi (semakin 0, semakin stabil modelnya)
```
